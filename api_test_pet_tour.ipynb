```python
import requests, pandas as pd

service_key = "c36f20fa8e010928512eff06c66f392b85ff6fc698dde757ce93564d10b03e97"  # Decoding 키 추천

base = "http://apis.data.go.kr/B551011/KorPetTourService2/areaBasedList2"
url = f"{base}?serviceKey={service_key}"

params = {
    "MobileOS": "ETC",
    "MobileApp": "ColabTest",
    "_type": "json",
    "numOfRows": 10,
    "pageNo": 1,
}

r = requests.get(url, params=params, timeout=20)

print("status:", r.status_code)
print("final url:", r.url)
print("content-type:", r.headers.get("Content-Type"))
print("preview:", r.text[:120])

# 200일 때만 JSON 파싱
if r.status_code == 200 and "json" in (r.headers.get("Content-Type") or ""):
    data = r.json()
    items = data["response"]["body"]["items"]["item"]
    df = pd.DataFrame(items)
    display(df[["title","addr1","tel"]].head(10))
else:
    print("❗인증/요청 문제. 위 final url/preview 확인!")
```

```python
import folium
from folium.plugins import MarkerCluster
import pandas as pd

# 좌표 컬럼을 숫자로 변환 (문자열로 오는 경우가 많음)
df2 = df.copy()
df2["mapx"] = pd.to_numeric(df2.get("mapx"), errors="coerce")
df2["mapy"] = pd.to_numeric(df2.get("mapy"), errors="coerce")

# 좌표 있는 것만
df2 = df2.dropna(subset=["mapx", "mapy"])

# 너무 많으면 느려서 200개만(원하면 숫자 바꿔도 됨)
df_plot = df2.head(50)

# 중심점(평균)으로 지도 시작
center_lat = df_plot["mapy"].mean()
center_lon = df_plot["mapx"].mean()

m = folium.Map(location=[center_lat, center_lon], zoom_start=7)

cluster = MarkerCluster().add_to(m)

for _, row in df_plot.iterrows():
    title = row.get("title", "")
    addr = row.get("addr1", "")
    tel = row.get("tel", "")

    popup = folium.Popup(f"<b>{title}</b><br>{addr}<br>{tel}", max_width=300)
    folium.Marker(
        location=[row["mapy"], row["mapx"]],
        popup=popup
    ).add_to(cluster)

m
```
